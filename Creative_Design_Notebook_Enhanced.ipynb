## 2. Interactive Story Generation

To make AI writing more interactive, the user can input a story prompt and select a writing style (e.g., **Fantasy, Thriller, Romance**). The AI will then generate a continuation of the story accordingly.

```
from transformers import pipeline

# Load GPT-2 pre-trained model
story_generator = pipeline('text-generation', model='gpt2')

# User input
user_story = input("Enter your story opening: ")
style = input("Choose a style (Fantasy/Thriller/Romance): ").lower()

# Style-specific prompts
style_prompts = {
    "fantasy": f"In a magical kingdom, {user_story}",
    "thriller": f"It was a dark and stormy night. Suddenly, {user_story}",
    "romance": f"Under the soft glow of the moonlight, {user_story}"
}

# Generate AI continuation
prompt = style_prompts.get(style, user_story)
generated_story = story_generator(prompt, max_length=100, num_return_sequences=1)

print("\n📝 AI-generated story:\n")
print(generated_story[0]['generated_text'])
```

## 3. Ethical Considerations in AI-Generated Writing

### 3.1 AI Bias and Copyright Issues
- AI-generated stories may reflect **biases from their training data**.
- **Case Study:** *CNET’s failed attempt at AI journalism*, where AI-generated news contained factual errors.
- **Copyright Issues:** *New York Times v. OpenAI* – News organizations claim AI models use their content unfairly.

### 3.2 Proposed Solutions
1. **Transparency**: Clearly mark AI-generated text as AI-assisted.
2. **Fair Compensation**: Authors should be credited if AI is trained on their work.
3. **Bias Detection**: Implement AI fairness testing before deploying AI-generated content.
