### Functions

```python
def greet(name):
    return 'Hello, ' + name + '!'

print(greet('World'))
```

### Loops

```python
for i in range(5):
    print(f'Hello, World! {i+1}')
```

### Lists

```python
greetings = ['Hello', 'Hi', 'Hey', 'Hola']
for g in greetings:
    print(g + ', World!')
```

Copilot Interaction Log
- **Prompt:** Create and loop through a list of greetings.
- **Copilot Response:** Suggested using a `for` loop over a list.
- **Reflection:** Lists store multiple items.

### Dictionaries

```python
translations = {'English': 'Hello', 'Spanish': 'Hola', 'French': 'Bonjour'}
for lang, greet in translations.items():
    print(f'In {lang}, we say {greet}!')
```

Copilot Interaction Log
- **Prompt:** Create a dictionary for greetings in different languages.
- **Copilot Response:** Suggested using `dict` and `.items()`.
- **Reflection:** Dictionaries store key-value pairs.

### Arrays

```python
import array
arr = array.array('i', [1, 2, 3, 4])
print(arr)
```

Copilot Interaction Log
- **Prompt:** How to create a simple integer array?
- **Copilot Response:** Suggested `array.array()`.
- **Reflection:** Arrays are useful for fixed-type lists.

### Conditional Statements

```python
name = input("What's your name? ")
if name == 'Alice':
    print('Hello, Alice!')
elif name == 'Bob':
    print('Hello, Bob!')
else:
    print(f'Hello, {name}!')
```

Copilot Interaction Log
- **Prompt:** How to write a conditional greeting program?
- **Copilot Response:** Suggested `if-elif-else`.
- **Reflection:** Conditions make programs interactive.

## 5. Using matplotlib to process image data

### Step1: Import matplotlib

```python
import matplotlib.pyplot as plt
```

### Step2: Load and display the image

Read and display images

```python
test_picture = plt.imread("picture1.jpg")
plt.imshow(test_picture)
plt.axis("off") 
plt.show()  
```

plt.imread("picture1.jpg"): Read the image file and save it in the variable test_picture.

plt.imshow(test_picture): Display the image.

plt.axis("off"): Hide the axis to keep the interface clean.

plt.show(): Execute the display command to let Notebook display the image.

### Step 3: View the image’s pixel data

```python
print(test_picture)
```

### Step4: Modify the image's colors

Adjust brightness (reduce by 1/3)

```python
test_picture_filtered = 2 * test_picture / 3
```

Display the modified image

```python
plt.imshow(test_picture_filtered)
plt.axis("off")
plt.show()
```

### Step5: Save the edited image

```python
test_picture_filtered = test_picture_filtered / 255.0
plt.imsave("filtered_picture.jpg", test_picture_filtered)
```
