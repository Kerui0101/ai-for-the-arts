## Hello World

```
print('Hello, World!')
```

## Variables

```
greeting = 'Hello'
name = 'World'
print(greeting + ', ' + name + '!')
```

## Functions

```
def greet(name):
    return 'Hello, ' + name + '!'

print(greet('World'))
```

## Loops

```
for i in range(5):
    print(f'Hello, World! {i+1}')
```

## Lists

```
greetings = ['Hello', 'Hi', 'Hey', 'Hola']
for g in greetings:
    print(g + ', World!')
```

## Dictionaries

```
translations = {'English': 'Hello', 'Spanish': 'Hola', 'French': 'Bonjour'}
for lang, greet in translations.items():
    print(f'In {lang}, we say {greet}!')
```

## Arrays

```
import array
arr = array.array('i', [1, 2, 3, 4])
print(arr)
```

## Conditional Statements

```
name = input("What's your name? ")
if name == 'Alice':
    print('Hello, Alice!')
elif name == 'Bob':
    print('Hello, Bob!')
else:
    print(f'Hello, {name}!')
```

## Operators

```
a = 5
b = 3
print(f'{a} + {b} = {a + b}')
print(f'{a} * {b} = {a * b}')
```

## Importing Libraries

```
import random
print(f'Random number: {random.randint(1, 10)}')
```

## Classes and Objects

```
class Greeter:
    def __init__(self, name):
        self.name = name
    def greet(self):
        return f'Hello, {self.name}!'

g = Greeter('World')
print(g.greet())
```

## Viewing the Image's Pixel Data

```
from PIL import Image
import numpy as np

img = Image.open('media/picture1.jpg')
img_array = np.array(img)
print('Image Shape:', img_array.shape)
print('Pixel Data Sample:', img_array[0, :5])
```

## Modify the Image's Colors

```
import matplotlib.pyplot as plt

filtered_img_array = img_array // 2
plt.imshow(filtered_img_array)
plt.title('Dimmed Image')
plt.show()
```

## Display the Modified Image

```
plt.imshow(filtered_img_array)
plt.title('Modified Image (Dimmed)')
plt.axis('off')
plt.show()
```

## Save the Edited Image

```
filtered_img = Image.fromarray(filtered_img_array.astype('uint8'))
filtered_img.save('media/filtered-sample-image.jpg')
```

## Exploring Datasets with scikit-learn and Pandas

```
import pandas as pd
from sklearn.datasets import load_iris

iris = load_iris()
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['species'] = [iris.target_names[i] for i in iris.target]
print(df.head())
```
